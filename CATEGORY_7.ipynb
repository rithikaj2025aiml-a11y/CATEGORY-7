```python
#1.Map Colouring using backtracking:
def is_safe(region, colour, assignment, graph):
    for neighbour in graph[region]:
        if assignment.get(neighbour) == colour:
            return False
    return True
def map_coloring(graph, colors):
    assignment = {}
    def backtrack():
        if len(assignment) == len(graph):
            return True
        region = next(r for r in graph if r not in assignment)
        for colour in colors:
            if is_safe(region, colour, assignment, graph):
                assignment[region] = colour
                if backtrack():
                    return True
                del assignment[region]
        return False
    if backtrack():
        return assignment
    return None
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}
colors = ['Red', 'Green', 'Blue']
solution = map_coloring(graph, colors)
print("Map Colouring Solution:")
print(solution)
```

```python
#2.Crypt-Arithmetic Puzzle:
from itertools import permutations
letters='SENDMORY'
digits=range(10)
for values in permutations(digits,len(letters)):
    S, E, N, D, M, O, R, Y=values
    if S==0 or M==0:
        continue
    SEND=1000*S+100*E+10*N +D
    MORE=1000*M+100*O+10*R+E
    MONEY=10000*M+1000*O+100*N+10*E+Y
    if SEND+MORE==MONEY:
        print("Solution found:")
        print("S =", S)
        print("E =", E)
        print("N =", N)
        print("D =", D)
        print("M =", M)
        print("O =", O)
        print("R =", R)
        print("Y =", Y)
        print("\nSEND =", SEND)
        print("MORE =", MORE)
        print("MONEY =", MONEY)
        break
```

```python
#3.Local Search for CSP:
import random
N = 8
def conflicts(board, row, col):
    count = 0
    for r in range(N):
        if r == row:
            continue
        c = board[r]
        if c == col:
            count += 1
        if abs(r - row) == abs(c - col):
            count += 1
    return count
def min_conflicts():
    board = [random.randint(0, N - 1) for _ in range(N)]
    for step in range(10000):
        bad_rows = [
            row for row in range(N)
            if conflicts(board, row, board[row]) > 0
        ]
        if not bad_rows:
            return board
        row = random.choice(bad_rows)
        min_value = min(
            conflicts(board, row, col)
            for col in range(N)
        )
        best_columns = [
            col for col in range(N)
            if conflicts(board, row, col) == min_value
        ]
        board[row] = random.choice(best_columns)

    return None
solution = min_conflicts()
print("8-Queens Solution:")
print(solution)
if solution:
    for row in range(N):
        for col in range(N):
            if solution[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
```
